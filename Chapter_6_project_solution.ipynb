## 2. Multiplication of matrices

Denote:

- $X$ — feature matrix (zero column consists of unities)

- $y$ — target vector

- $P$ — matrix by which the features are multiplied

- $w$ — linear regression weight vector (zero element is equal to the shift)

Predictions:

$$
a = Xw
$$

Training objective:

$$
\min_w d_2(Xw, y)
$$

Training formula:

$$
w = (X^T X)^{-1} X^T y
$$

**Answer:** linear regression quality will not change.

**Justification.**

Denote $w_X$ — Linear regression parameters that are obtained during training using the feature matrix X. That is, $\min_w d_2(Xw, y)$ is obtained at $w_X$.

Transformed task: $\min_w d_2(XPw, y)$

Introduce vector $w_P = P^{-1} w_X$ and put it in the transformed task:

$$
d_2(XPw_P, y) = d_2(XPP^{-1} w_X, y) = d_2(Xw_X, y)
$$

We obtained a one-to-one correspondence between the parameters of the original and transformed tasks. So the minimum will be achieved at the same value of the loss function.

## 3. Transformation algorithm

**Algorithm**

- Create matrix $P$ with normally distributed random numbers
- Check that it is invertible. If not, repeat the generation until we get an invertible matrix..
- Multiply  feature matrix $X$ by matrix $P$

**Justification**

We will use the previous paragraph: when multiplying the features by an invertible matrix, the quality of the linear regression will not change.
But multiplying by random numbers will make the initial values unrecoverable.

## 4. Algorithm test

```python
from sklearn.model_selection import train_test_split

features = data.drop('insurance benefits', axis=1).values
target = data['insurance benefits'].values

features_train, features_test, target_train, target_test = train_test_split(
    features, target, test_size=0.2, random_state=12345)
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score

model = LinearRegression()
model.fit(features_train, target_train)
pred_train = model.predict(features_train)
pred_test = model.predict(features_test)

print("Train")
print("MSE =", mean_squared_error(target_train, pred_train))
print("R2 =", r2_score(target_train, pred_train))
print()

print("Test")
print("MSE =", mean_squared_error(target_test, pred_test))
print("R2 =", r2_score(target_test, pred_test))
```

```python
import numpy as np

state = np.random.RandomState(12345)
transform_matrix = state.normal(size=(features.shape[1], features.shape[1]))
rev_matrix = np.linalg.inv(transform_matrix)


new_features_train = features_train.dot(transform_matrix)
new_features_test = features_test.dot(transform_matrix)
```
