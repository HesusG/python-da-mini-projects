## Agregar datos por usuario

Agregamos los datos por usuario por periodo para poder tener solo un registro por usuario por periodo con relación a llamadas, mensajes e internet.

```python
# Calcula el número de llamadas hechas por cada usuario por mes. Guarda el resultado.
calls_month = calls.groupby(['user_id','month'])['id'].count().reset_index()

calls_month.head()
```

```python
calls_month
```

```python
# Calcula la cantidad de minutos que usó cada usuario por mes. Guarda el resultado.
min_month = calls.groupby(['user_id','month'])['duration_round'].sum().reset_index()
min_month.head()
```

```python
# Calcula el número de mensajes enviados por cada usuario por mes. Guarda el resultado.
sms_month = messages.groupby(['user_id','month'])['id'].count().reset_index()

sms_month.head()

```

```python
sms_month
```

```python
# Calcula el volumen del trafico de internet usado por cada usuario por mes. Guarda el resultado.
mb_month = internet.groupby(['user_id','month'])['mb_used'].sum().reset_index()

#Redondeamos el consumo mensual 1gb = 1024 mb, 
#el redondeo es por mes, no individual como las llamadas.

mb_month['gb_round'] = mb_month['mb_used'] / 1024
mb_month['gb_round'] = np.ceil(mb_month['gb_round'])
mb_month.head()
```

Ahora juntamos las tablas en un dataframe para que un registro en él represente lo que un único usuario consumió en un mes determinado.

```python
# Combina los datos para las llamadas, minutos, mensajes e internet con base en el user_id y el mes
resumen = pd.merge(calls_month, min_month, on=['user_id','month'], how='outer')  
resumen = pd.merge(resumen, sms_month, on=['user_id','month'], how='outer')
resumen = pd.merge(resumen, mb_month, on=['user_id','month'], how='outer')

# Renombramos las columas para cantidad de mensajes y llamadas
resumen = resumen.rename(columns={'id_x': 'n_calls'})
resumen = resumen.rename(columns={'id_y': 'n_sms'}) 
print(resumen.isna().sum())
resumen.head()
```

Producto de la unión podemos observar que aparecen valores ausentes, ésto se debe a que por ejemplo hubo usuarios que no realizaron llamadas en cierto mes o mensajes de texto, fecha de subscripción, etc.

```python
# Dejamos los NaN en 0 (para poder realizar las operaciones correspondientes)

#columns_to_replace = ['n_calls', 'duration_round', 'n_sms', 'mb_used', 'gb_round']

#for value in columns_to_replace:
#    resumen[value] = resumen[value].fillna(0)

resumen = resumen.fillna(0)

print(resumen.isna().sum())
resumen.head()
```

```python
# Añade la información del plan
users_2 = users.loc[:,['user_id','plan','city']]
users_2 = users_2.rename(columns={'plan': 'plan_name'}) 

print(users_2.head())
resumen = users_2.merge(resumen, on=['user_id'], how='right').merge(plans, on='plan_name', how='right')

#Agregamos una columna para dejar los megas inluídos en el plan en gigas

resumen['gb_per_month_included'] = resumen['mb_per_month_included'] / 1024

resumen.tail(60)
```

Ahora calcularemos los ingresos mensuales por cada usuario

```python
# Calcula el ingreso mensual para cada usuario
#Llamadas
resumen['cargo_calls'] = (resumen['duration_round'] - resumen['minutes_included']) * resumen['usd_per_minute'] 

#Corregimos los números negativos dejándolos en 0 ya que no sobrepasaron los límites del plan
resumen['cargo_calls'] = np.where(resumen['cargo_calls'] < 0, 0, resumen['cargo_calls'])


#SMS
resumen['cargo_sms'] = (resumen['n_sms'] - resumen['messages_included']) * resumen['usd_per_message'] 

#Corregimos los números negativos dejándolos en 0 ya que no sobrepasaron los límites del plan
resumen['cargo_sms'] = np.where(resumen['cargo_sms'] < 0, 0, resumen['cargo_sms'])


#Internet
resumen['cargo_internet'] = (resumen['gb_round'] - resumen['gb_per_month_included']) * resumen['usd_per_gb']


#Corregimos los números negativos dejándolos en 0 ya que no sobrepasaron los límites del plan
resumen['cargo_internet'] = np.where(resumen['cargo_internet'] < 0, 0, resumen['cargo_internet'])

#Agregamos una columna con el ingreso total
resumen['total_ingreso'] = resumen['usd_monthly_pay'] + resumen['cargo_calls'] + resumen['cargo_sms'] + resumen['cargo_internet']

resumen.head()
```

### Llamadas

Comparamos la duración promedio de llamadas por cada plan por cada mes.

```python
# Compara la duración promedio de llamadas por cada plan por cada mes. Dibuja una gráfica de barras para visualizarlo.
compare_calls_duration = resumen.groupby(['month','plan_name'])['duration_round'].mean().reset_index()
fig, ax = plt.subplots()

#Gráfico

plan_surf_ll = compare_calls_duration[compare_calls_duration['plan_name']== 'surf']
ax.bar(plan_surf_ll['month'], plan_surf_ll['duration_round'], label='Surf', color='blue', alpha=0.5)

plan_ultimate_ll = compare_calls_duration[compare_calls_duration['plan_name']== 'ultimate']
ax.bar(plan_ultimate_ll['month'], plan_ultimate_ll['duration_round'], label='Ultimate', color='yellow', alpha=0.5)

ax.set_xlabel('MES')
ax.set_ylabel('DURACION PROMEDIO (min)')
ax.set_title('DURACIÓN PROMEDIO DE LLAMADAS POR CADA PLAN Y CADA MES')
ax.legend()
plt.show()

#Gráfico barras separadas
compare_calls_duration_2 = resumen.pivot_table(index='month', columns='plan_name', values='duration_round')

compare_calls_duration_2.plot(kind='bar', figsize=(6,4), grid=True, title='Duración promedio de llamadas por cada plan y cada mes', xlabel='Mes', ylabel='Duración promedio(min)')
```
